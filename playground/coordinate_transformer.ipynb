```python
from pyproj import Transformer
```

```python
proj_katec = "+proj=tmerc +lat_0=38 +lon_0=128 +k=0.9999 +x_0=400000 +y_0=600000 +ellps=bessel +units=m +no_defs"
transformer = Transformer.from_crs("epsg:4326", proj_katec, always_xy=True)

songpa_lat, songpa_lon = 37.51447, 127.10595
katec_x, katec_y = transformer.transform(songpa_lon, songpa_lat)
print(katec_x , katec_y)
```

```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
API_KEY = os.getenv('OPINET_API_KEY')
```

```python
url = "https://www.opinet.co.kr/api/aroundAll.do"
params = {
    "code": API_KEY,
    "out": "json",
    "x": katec_x,
    "y": katec_y,
    "radius": 5000,
    "prodcd": "B027" , #"B027",  # 휘발유 기준 (경유는 D047)
    "sort": 1          # 1: 가격순, 2: 거리순
}
```

```python
response = requests.get(url, params=params)
data = response.json()
oil_list = data.get('RESULT', {}).get('OIL', [])
display(oil_list)
```

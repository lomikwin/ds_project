```python
import os
import boto3
from botocore.client import Config
from dotenv import load_dotenv ,find_dotenv
# 1. 설정 로드
load_dotenv(find_dotenv)
endpoint = os.getenv('MINIO_ENDPOINT')
access = os.getenv('MINIO_ACCESS_KEY')
secret = os.getenv('MINIO_SECRET_KEY')
```

```python

# 2. MinIO 연결 설정
s3 = boto3.resource('s3',
    endpoint_url=f"http://{endpoint}",
    aws_access_key_id=access,
    aws_secret_access_key=secret,
    config=Config(signature_version='s3v4'),
    region_name='us-east-1'
)
```

```python

# [작업 1] petroleum-project 내의 sample_data.parquet 삭제
try:
    s3.Object('petroleum-project', 'sample_data.parquet').delete()
    print("✅ sample_data.parquet 삭제 완료")
except Exception as e:
    print(f"❌ 파일 삭제 실패: {e}")
```

```python

# [작업 2] test-bucket 삭제 (내용물까지 싹 비우고 삭제)
try:
    bucket = s3.Bucket('test-bucket')
    bucket.objects.all().delete() # 내용물 먼저 삭제
    bucket.delete()               # 버킷 삭제
    print("✅ test-bucket 삭제 완료")
except Exception as e:
    print(f"❌ 버킷 삭제 실패: {e}")
```
